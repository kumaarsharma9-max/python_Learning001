**Loops in python**

**Loops in Python are used to repeat actions efficiently. The main types are For loops (counting through items) and While loops (based on conditions).**


**For Loop**<br>
**For loop is used to iterate over a sequence such as a list,tuple, string or range.**<br>
**it allow to execute a block of code repeatedly, once for each item in the sequence**

```python
n = 4
for i in range(0, n):
    print(i)
```

```python
n = 4
for i in range(0, n):
  print(i)
```

```python
list_ = ['apple','banana','orange']
for i in list_:
  print(i)
```

```python
tup = ('Time','place','space')
for i in tup:
  print(i)
```

```python
s = 'abc'
for x in s:
  print(x)
```

```python
dic = dict({'x':123,'y':345})
for x in dic:
  print("%s %d" % (x,dic[x]))
```

```python
set = {10,10,20,30}
for x in set:
  print(x)
```

**While Loop**

**while loop is used to execute a block of statements repeatedly until a given condition is satisfied. When the condition becomes false, the line immediately after the loop in the program is executed.**

```python
cnt = 0
while (cnt < 3):
    cnt = cnt + 1
    print("Hello Geek")
```

```python
count = 0
while (count < 9):
  count= count + 1
  print("Hello , Kumaar")
```

**Infinite While Loop**<br>
To execute a block of code infinite number of times we can use the while loop.

Code given below uses a 'while' loop with the condition "True", which means that the loop will run infinitely until we break out of it using "break" keyword or some other logic.

```python
while (True):
    print("Hello Geek")
```

```python

for i in range(1, 5):
    for j in range(i):
        print(i, end=' ')
    print()
```

```python
for i in range(1,5):
  for j in range(i):
    print(i,end=' ')
  print()
```

**Loop Control Statements**

**Loop control statements change execution from their normal sequence.**<br>
**When execution leaves a scope, all automatic objects that were created in that scope are destroyed. Python supports the following control statements.**

**Continue Statement**

**The continue statement in Python returns the control to the beginning of the loop**

```python
for letter in 'KumaarSharma':
    if letter == 'a' or letter == 'r':
        continue
    print('Current Letter :', letter)
```

**break statement**

**The break statement in Python brings control out of the loop.**

```python
for letter in 'Kumaarsharma':
    if letter == 'e' or letter == 's':
        break

print('Current Letter :', letter)
```

**Pass statement**

**Pass Statement**<br>
**We use pass statement in Python to write empty loops. Pass is also used for empty control statements, functions and classes.**

```python
for letter in 'KumaarSharma':
    pass
print('Last Letter :', letter)
```
